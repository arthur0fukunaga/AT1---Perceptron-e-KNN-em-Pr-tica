```python
import numpy as np

# Desafio 1: Classificação Binária com Perceptron Treinável

X_treino_fraude = np.array([
    [1.5, 1.0], [2.0, 2.0], [3.5, 1.5], [3.0, 3.0],
    [6.5, 5.0], [7.0, 7.0], [8.5, 6.0], [9.0, 8.0]
])
y_treino_fraude = np.array([0, 0, 0, 0, 1, 1, 1, 1])

transacao_A = np.array([5.0, 4.0])
transacao_B = np.array([7.0, 6.5])

def treinar_perceptron(X, y, lr=0.1, epocas=100):
    n_amostras, n_features = X.shape
    pesos = np.ones(n_features, dtype=float)
    vies = 1.0

    for _ in range(epocas):
        erros = 0
        for i in range(n_amostras):
            z = np.dot(X[i], pesos) + vies
            y_pred = 1 if z >= 0 else 0
            erro = y[i] - y_pred
            if erro != 0:
                pesos += lr * erro * X[i]
                vies += lr * erro
                erros += 1
        if erros == 0:
            break
    return pesos, vies

def predizer_perceptron(X_amostra, pesos, vies):
    z = np.dot(X_amostra, pesos) + vies
    return 1 if z >= 0 else 0

pesos_calibrados, vies_calibrado = treinar_perceptron(X_treino_fraude, y_treino_fraude)

print("=" * 60)
print("DESAFIO 1: PERCEPTRON")
print("=" * 60)
print(f"Pesos Finais: {pesos_calibrados}")
print(f"Viés Final: {vies_calibrado:.4f}")
print(f"Transação A {transacao_A}: Classe {predizer_perceptron(transacao_A, pesos_calibrados, vies_calibrado)}")
print(f"Transação B {transacao_B}: Classe {predizer_perceptron(transacao_B, pesos_calibrados, vies_calibrado)}\n")


# Desafio 2: Predição de Risco de Churn com Classificador KNN

X_treino_churn = np.array([
    [2.0, 1.0], [3.0, 0.0], [5.0, 1.0], [6.0, 2.0],
    [15.0, 4.0], [18.0, 5.0], [20.0, 4.0], [22.0, 6.0]
])
y_treino_churn = np.array([0, 0, 0, 0, 1, 1, 1, 1])

cliente_1 = np.array([13.0, 10.0])
cliente_2 = np.array([7.0, 1.5])

def classificar_knn(X_treino, y_treino, ponto_consulta, k=3, metrica='euclidiana'):
    if metrica == 'euclidiana':
        distancias = np.sqrt(np.sum((X_treino - ponto_consulta) ** 2, axis=1))
    elif metrica == 'manhattan':
        distancias = np.sum(np.abs(X_treino - ponto_consulta), axis=1)

    indices_k = np.argsort(distancias)[:k]
    distancias_k = distancias[indices_k]
    classes_k = y_treino[indices_k]

    valores, contagens = np.unique(classes_k, return_counts=True)
    classe_predita = valores[np.argmax(contagens)]

    return classe_predita, indices_k, distancias_k

classes_map_churn = {0: "Baixo Risco", 1: "Alto Risco"}

print("=" * 60)
print("DESAFIO 2: CLASSIFICADOR KNN")
print("=" * 60)

for cliente, k, nome in [(cliente_1, 3, "Cliente 1"), (cliente_2, 5, "Cliente 2")]:
    classe, idxs, dists = classificar_knn(X_treino_churn, y_treino_churn, cliente, k=k)
    print(f"\nDIAGNÓSTICO DO {nome} (K={k}):")
    print(f"  • Classe Predita: {classe} ({classes_map_churn[classe]})")
    print(f"  • Índices dos Vizinhos: {idxs}")
    print(f"  • Distâncias: {np.round(dists, 4)}")


# Desafio 3: Recomendação de Servidores Cloud por Similaridade Espacial

catalogo_servidores = np.array([
    [2.0, 4.0, 50.0],
    [4.0, 8.0, 100.0],
    [8.0, 16.0, 250.0],
    [16.0, 32.0, 500.0],
    [32.0, 64.0, 1000.0],
    [64.0, 128.0, 2000.0]
])

nomes_servidores = [
    "Micro Instância Web",
    "Standard App Server",
    "Medium Backend & Cache",
    "Database Enterprise",
    "High Performance Computing",
    "Big Data & AI Training"
]

perfil_demandado = np.array([12.0, 28.0, 850.0])

def recomendar_servidores(catalogo, demanda, k=2):
    distancias = np.sqrt(np.sum((catalogo - demanda) ** 2, axis=1))
    indices_ordenados = np.argsort(distancias)[:k]
    return indices_ordenados, distancias[indices_ordenados]

indices_rec, dists_rec = recomendar_servidores(catalogo_servidores, perfil_demandado, k=2)

print("\n" + "=" * 60)
print("DESAFIO 3: RECOMENDADOR DE SERVIDORES CLOUD")
print("=" * 60)
for rank, (idx, dist) in enumerate(zip(indices_rec, dists_rec), start=1):
    servidor = catalogo_servidores[idx]
    print(f"{rank}º Lugar: {nomes_servidores[idx]}")
    print(f"   Configuração: {servidor[0]:.0f} vCPUs, {servidor[1]:.0f} GB RAM, {servidor[2]:.0f} GB SSD")
    print(f"   Distância Geométrica: {dist:.4f}")
```
